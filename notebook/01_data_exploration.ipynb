# 01 — Data Exploration

Explore and validate the 2016–2025 Kiambu weather dataset before irrigation modelling.

## 1. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option("display.max_columns", None)
```

## 2. Load data

```python
DATA_PATH = Path("../data/raw/kiambu_weather.csv")
weather_df = pd.read_csv(DATA_PATH)
weather_df["Date"] = pd.to_datetime(weather_df["Date"])

print("Shape:", weather_df.shape)
display(weather_df.head())
```

## 3. Data quality checks

```python
print("Columns:", weather_df.columns.tolist())
print("\nMissing values:\n", weather_df.isna().sum())
print("\nDuplicate dates:", weather_df["Date"].duplicated().sum())
print("\nDate range:", weather_df["Date"].min(), "to", weather_df["Date"].max())
print("\nDtypes:\n", weather_df.dtypes)
```

## 4. Derived temperature feature

```python
weather_df["Tmean"] = (
    weather_df["Max_Temperature"] + weather_df["Min_Temperature"]
) / 2
display(weather_df[["Date","Max_Temperature","Min_Temperature","Tmean"]].head())
```

## 5. Descriptive statistics

```python
display(weather_df.describe().T)
```

## 6. Daily weather trends

```python
fig, axes = plt.subplots(4, 1, figsize=(12, 12), sharex=True)

axes[0].plot(weather_df["Date"], weather_df["Tmean"])
axes[0].set_ylabel("Tmean (°C)")

axes[1].plot(weather_df["Date"], weather_df["Rainfall"])
axes[1].set_ylabel("Rainfall (mm)")

axes[2].plot(weather_df["Date"], weather_df["Solar_Radiation"])
axes[2].set_ylabel("Solar radiation")

axes[3].plot(weather_df["Date"], weather_df["Relative_Humidity"])
axes[3].set_ylabel("RH (%)")
axes[3].set_xlabel("Date")

plt.tight_layout()
plt.show()
```

## 7. Correlation

```python
numeric = [
    "Temperature","Max_Temperature","Min_Temperature",
    "Relative_Humidity","Rainfall","Wind_Speed",
    "Solar_Radiation","Tmean"
]
display(weather_df[numeric].corr().round(2))
```

## 8. Time continuity

```python
expected = pd.date_range(weather_df["Date"].min(), weather_df["Date"].max(), freq="D")
missing_dates = expected.difference(weather_df["Date"])
print("Expected days:", len(expected))
print("Observed days:", len(weather_df))
print("Missing dates:", len(missing_dates))
```
